# 对应分析 CA

从固定种子生成列联表。CA 以独立性期望为基准，对质量加权残差矩阵做 SVD；奇异值平方是惯量，不应与 PCA 结果混同。

```python
import numpy as np
SEED = 6801  # fixed synthetic seed
rng = np.random.default_rng(SEED)
rate = np.array([[42,22,8,5,3],[35,28,9,4,3],[8,12,38,20,8],[6,10,32,25,12],[5,6,12,36,28],[3,7,10,28,40]], dtype=float)
table = rng.poisson(rate); table[table < 1] = 1
P = table / table.sum(); r, c = P.sum(axis=1), P.sum(axis=0)
S = (P - np.outer(r, c)) / np.sqrt(np.outer(r, c))
U, singular_values, Vt = np.linalg.svd(S, full_matrices=False)
(singular_values ** 2)[:2]
```

```python
row_coordinates = U[:, :2] * singular_values[:2] / np.sqrt(r[:, None])
column_coordinates = Vt.T[:, :2] * singular_values[:2] / np.sqrt(c[:, None])
row_coordinates.shape, column_coordinates.shape
```
