```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt              #加载基本绘图包
plt.rcParams['font.sans-serif']=['Heiti TC'];  #设置中文字体为黑体

from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"
```

```python
# 读取数据
epu = pd.read_excel(f'All_Country_Data.xlsx')

epu['DATE'] = pd.to_datetime(epu[['Year', 'Month']].assign(DAY=1))
epu.set_index('DATE', inplace=True)
epu.index = pd.to_datetime(epu.index,format='%Y%m%d')
epu = epu.loc['2000-01-01':]
```

```python
index_df = pd.read_excel('上证和科创综指.xls')
index_pivot = index_df.pivot_table(columns= '指数名称_IdxNm', index='交易日期_TrdDt', values='收盘价(元/点)_ClPr')
SH_index = index_pivot['上证指数']
SH_index.index = pd.to_datetime(SH_index.index,format='%Y%m%d')
SH_index = SH_index.loc['2000-01-01':]
```

```python
epu_china = epu['Mainland China'];print(f'''epu_china描述性统计: 
                                        {epu_china.describe()}''')
```

```python
def standardize(x):
    mean = x.mean()
    std = x.std()
    standardized_series = (x - mean) / std
    return standardized_series
def normalize(x): 
    z=(x-x.min())/(x.max()-x.min())*600
    return(z)
def normalize_300(x): 
    z=(x-x.min())/(x.max()-x.min())*600+300
    return(z)
```

```python
epu_china_standardized = standardize(epu_china);
epu_china_normalized = normalize(epu_china);
epu_china_normalized_300 = normalize_300(epu_china);
```

```python
fig, axs = plt.subplots(2, 2, figsize=(14, 10));

axs[0, 0].plot(epu_china_standardized, label='标准化');
axs[0, 0].set_title('中国政治经济不确定性指数 (标准化)');
axs[0, 0].set_xlabel('日期');
axs[0, 0].set_ylabel('标准化指数值');

axs[0, 1].plot(epu_china_normalized, label='归一化');
axs[0, 1].set_title('中国政治经济不确定性指数 (归一化)');
axs[0, 1].set_xlabel('日期');
axs[0, 1].set_ylabel('标准化指数值');

axs[1, 0].plot(epu_china_normalized_300, label='归一化后纵轴+300');
axs[1, 0].set_title('中国政治经济不确定性指数 (归一化后纵轴+300)');
axs[1, 0].set_xlabel('日期');
axs[1, 0].set_ylabel('标准化指数值');

# 隐藏右下角的子图
axs[1, 1].axis('off');

# 调整布局
plt.tight_layout();
plt.show();
```
