# 多指标综合评价：熵权 TOPSIS

先声明指标方向：收入、就业、创新、服务为正向，污染和负债为逆向。方向处理后进行极差标准化，再用熵权法赋权和 TOPSIS 排序；权重决定排序，需做敏感性分析。

```python
import numpy as np
import pandas as pd
SEED = 6901  # fixed synthetic seed
rng = np.random.default_rng(SEED)
columns = ['income','employment','innovation','pollution','debt','service']
X = pd.DataFrame(np.column_stack([rng.normal(65,8,12), rng.normal(72,7,12), rng.normal(55,10,12), rng.normal(38,6,12), rng.normal(48,9,12), rng.normal(60,8,12)]), columns=columns)
cost = {'pollution', 'debt'}
Z = X.copy()
for col in columns:
    lo, hi = X[col].min(), X[col].max()
    Z[col] = ((hi - X[col]) if col in cost else (X[col] - lo)) / (hi - lo)
P = Z / Z.sum(axis=0)
entropy = -(P * np.log(P.clip(lower=np.finfo(float).eps))).sum(axis=0) / np.log(len(X))
weights = (1 - entropy) / (1 - entropy).sum()
weights
```

```python
weighted = Z * weights
best, worst = weighted.max(), weighted.min()
d_best = np.sqrt(((weighted - best) ** 2).sum(axis=1))
d_worst = np.sqrt(((weighted - worst) ** 2).sum(axis=1))
scores = d_worst / (d_best + d_worst)
scores.sort_values(ascending=False).head()
```
