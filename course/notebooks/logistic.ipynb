# Logistic 回归

固定种子生成二分类数据。分层留出测试集，并将标准化和 Logistic 模型封装在 Pipeline 中，避免用测试分布拟合预处理。

```python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, roc_auc_score
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
SEED = 6601  # fixed synthetic seed
X, y = make_classification(n_samples=260, n_features=6, n_informative=4, n_redundant=1, weights=[.58,.42], class_sep=1.25, random_state=SEED)
train, test = train_test_split(np.arange(len(y)), test_size=.25, random_state=SEED, stratify=y)
model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=2000, random_state=SEED)).fit(X[train], y[train])
prob = model.predict_proba(X[test])[:, 1]
roc_auc_score(y[test], prob), confusion_matrix(y[test], model.predict(X[test]), labels=[0, 1])
```
