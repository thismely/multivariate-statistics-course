# 判别分析：LDA 与 QDA

这是一个三分类模拟任务。分层切分后，将 StandardScaler 放在 Pipeline 内，只在训练行拟合；最终报告独立测试集的准确率和混淆矩阵。

```python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis, QuadraticDiscriminantAnalysis
from sklearn.metrics import accuracy_score, confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
SEED = 6401  # fixed synthetic seed
X, y = make_classification(n_samples=240, n_features=6, n_informative=4, n_redundant=1, n_classes=3, n_clusters_per_class=1, class_sep=1.7, random_state=SEED)
train, test = train_test_split(np.arange(len(y)), test_size=.25, random_state=SEED, stratify=y)
lda = make_pipeline(StandardScaler(), LinearDiscriminantAnalysis()).fit(X[train], y[train])
qda = make_pipeline(StandardScaler(), QuadraticDiscriminantAnalysis(reg_param=.05)).fit(X[train], y[train])
lda_pred = lda.predict(X[test]); qda_pred = qda.predict(X[test])
```

```python
accuracy_score(y[test], lda_pred), confusion_matrix(y[test], lda_pred, labels=[0, 1, 2])
```
