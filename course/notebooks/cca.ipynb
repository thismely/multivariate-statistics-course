# 典型相关分析 CCA

CCA 连接 X 与 Y 两组变量。两个 StandardScaler 分别在训练行拟合，CCA 使用已标准化的数据；训练集相关与测试集相关分开汇报。

```python
import numpy as np
from sklearn.cross_decomposition import CCA
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
SEED = 6701  # fixed synthetic seed
rng = np.random.default_rng(SEED)
latent = rng.normal(size=(240, 2))
X = latent @ np.array([[.9,.1],[.7,.3],[.2,.8],[-.1,.7]]).T + rng.normal(scale=.35, size=(240,4))
Y = latent @ np.array([[.8,.2],[.5,.6],[.1,.9]]).T + rng.normal(scale=.35, size=(240,3))
train, test = train_test_split(np.arange(240), test_size=.25, random_state=SEED)
sx, sy = StandardScaler().fit(X[train]), StandardScaler().fit(Y[train])
cca = CCA(n_components=2, scale=False, max_iter=2000).fit(sx.transform(X[train]), sy.transform(Y[train]))
```

```python
xc, yc = cca.transform(sx.transform(X[test]), sy.transform(Y[test]))
[abs(np.corrcoef(xc[:, i], yc[:, i])[0, 1]) for i in range(2)]
```
