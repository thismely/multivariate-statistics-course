# 因子分析 FA

固定种子生成六个观测变量。FA 估计公共方差与特殊方差，Varimax 旋转帮助解释载荷结构；PCA 追求总方差的线性表示，两者目标不同。

```python
import numpy as np
from sklearn.decomposition import FactorAnalysis
from sklearn.preprocessing import StandardScaler
SEED = 6201  # fixed synthetic seed
rng = np.random.default_rng(SEED)
F = rng.normal(size=(220, 2))
L = np.array([[.82,.08],[.76,.20],[.70,-.12],[.12,.86],[.26,.78],[-.16,.72]])
X = F @ L.T + rng.normal(scale=.35, size=(220, 6))
Xz = StandardScaler().fit_transform(X)
fa = FactorAnalysis(n_components=2, rotation='varimax', random_state=SEED).fit(Xz)
fa.components_.T
```

```python
loadings = fa.components_.T
communalities = (loadings ** 2).sum(axis=1)
uniqueness = fa.noise_variance_
communalities, uniqueness
```
