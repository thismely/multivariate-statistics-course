# 聚类分析：K-means 与 Ward

数据由固定种子生成。标准化后分别运行 K-means 与 Ward 层次聚类，用轮廓系数评价分离度；生成器标签只用于教学验证，不是现实类别。

```python
import numpy as np
from sklearn.cluster import AgglomerativeClustering, KMeans
from sklearn.datasets import make_blobs
from sklearn.metrics import adjusted_rand_score, silhouette_score
from sklearn.preprocessing import StandardScaler
SEED = 6301  # fixed synthetic seed
centers = np.array([[-2.8,-1.7,0,.8],[0,2.5,2,-1.8],[3,-.4,-2.3,1.8]])
X, truth = make_blobs(n_samples=180, centers=centers, cluster_std=.72, random_state=SEED)
Xz = StandardScaler().fit_transform(X)
kmeans_labels = KMeans(n_clusters=3, n_init=20, random_state=SEED).fit_predict(Xz)
ward_labels = AgglomerativeClustering(n_clusters=3, linkage='ward').fit_predict(Xz)
silhouette_score(Xz, kmeans_labels), adjusted_rand_score(truth, kmeans_labels)
```

```python
adjusted_rand_score(truth, ward_labels), silhouette_score(Xz, ward_labels)
```
